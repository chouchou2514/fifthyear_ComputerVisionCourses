4. Appliquer une ouverture 3 × 3 puis une fermeture 3 × 3. Nommer le résultat BW_net. Observer les
fragments, les trous et les contacts. Tester ensuite une fermeture 9 × 9 : quelles régions fusionnent ?

```python
#ouverture 3x3 puis fermeture 3x3
BW_net = ndi.binary_closing(ndi.binary_opening(BW, structure=SE_3), structure=SE_3)

BW_fermeture_9 = ndi.binary_closing(BW, structure=SE_9)

# Affichage des résultats
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(BW, cmap="gray")
plt.title("Image originale (BW)")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(BW_net, cmap="gray")
plt.title("BW_net (Ouverture 3x3 + Fermeture 3x3)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(BW_fermeture_9, cmap="gray")
plt.title("Fermeture 9x9")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Le combiné Ouverture + Fermeture (appelé filtre morphologique alternatif) permet d'assainir l'image binarisée sans altérer la taille globale des cellules :

- Fragments (Bruit de fond) : L'ouverture $3 \times 3$ supprime le bruit impulsionnel (« poivre et sel »), c'est-à-dire les petits points blancs isolés dans le fond de taille inférieure à l'élément structurant.

- Trous (Intérieur des cellules) : La fermeture $3 \times 3$ rebouche les petits trous noirs ou imperfections à l'intérieur de la surface des cellules.Contacts et contours : Les contours des objets sont lissés. Les cellules distinctes restent bien séparées car l'élément structurant de $3 \times 3$ est trop petit pour pontrer les espacements intercellulaires.

Bilan pour $BW\_net$ : On obtient un masque binaire propre, débarrassé du bruit de fond et des cavités internes, adapté pour l'étape de segmentation.

Effet de la Fermeture 9×9 et fusion des régionsEn augmentant la taille de l'élément structurant à $9 \times 9$, l'opération de fermeture devient beaucoup plus agressive :

- Fusion des cellules proches : La fermeture comble tous les espaces sombres d'une largeur inférieure ou égale à 9 pixels. Par conséquent, les cellules contiguës ou proches fusionnent pour ne former qu'un seul composant connexe.

- Disparition des contours individuels : On perd la séparation entre les objets adjacents, ce qui empêche le comptage précis ou la caractérisation individuelle des cellules.

Conclusion : Un élément structurant $3 \times 3$ est idéal pour le nettoyage. Une fermeture de taille $9 \times 9$ est excessive si l'objectif est d'isoler des cellules, mais s'avère utile si l'on cherche à délimiter un amas cellulaire ou une zone d'intérêt globale.

2. Comparer les deux résultats. Repérer les régions supprimées, séparées ou fusionnées par la

L'image d'origine $BW$ est « bruitée » et imparfaite. Les 270 composantes comptées au départ ne sont pas 270 vraies cellules :

- Le bruit de fond (artefacts) : Avant le nettoyage, l'image contient plein de tout petits points blancs isolés dans le fond (de la poussière, du bruit numérique). L'algorithme d'étiquetage prend chaque petit point isolé pour une cellule à part entière. L'ouverture $3 \times 3$ a supprimé tous ces faux objets.

- La fragmentation des cellules (les trous) : Certaines cellules réelles ont des petites zones sombres ou des trous au milieu. Sans la fermeture, l'algorithme voyait parfois une seule cellule comme 2 ou 3 morceaux séparés (donc plusieurs étiquettes pour un seul objet). La fermeture $3 \times 3$ a rebouché ces trous pour réunir ces morceaux en un seul objet.

270 est un sur-comptage évident : Il inclut le bruit + les morceaux de cellules fragmentées.

169 est une estimation beaucoup plus proche de la réalité : C'est le nombre de vraies cellules (ou d'amas de cellules) débarrassées des artefacts.

3. Calculer l’aire de chaque composante de BW_net, en pixels. Afficher ces aires et repérer la plus
grande région.

```python
#Fonction : aires = np.bincount(L.ravel())[1:]
'''
- L.ravel() :
    Convertit la matrice 2D des étiquettes L en un vecteur à 1 dimension (une simple liste plate de tous les pixels).

- np.bincount(...) :
    Compte le nombre d'occurrences de chaque entier présent dans le vecteur.

    L'élément à l'indice 0 donne le nombre de pixels du fond (label 0).

    L'élément à l'indice 1 donne le nombre de pixels de la région 1 (son aire en pixels).

    L'élément à l'indice 2 donne le nombre de pixels de la région 2, et ainsi de suite.

- [1:] :
    Extrait les résultats à partir de l'indice 1 jusqu'à la fin, ce qui exclut le fond (l'indice 0) pour ne conserver que les aires des vraies composantes/cellules.
'''
```

```python
aires_BW_net = np.bincount(L_net.ravel())[1:]

#Informations statistiques dans la console
print(f"Nombre de régions (BW_net) : {len(aires_BW_net)}")
print(f"Aire minimale : {aires_BW_net.min()} pixels")
print(f"Aire maximale : {aires_BW_net.max()} pixels")
print(f"Aire moyenne  : {aires_BW_net.mean():.2f} pixels")
```

```python
#Affichage de l'histogramme de distribution des aires (pour BW_net)
plt.figure(figsize=(8, 5))
plt.hist(aires_BW_net, bins=30, color="skyblue", edgecolor="black")
plt.title("Distribution de la taille des cellules (BW_net)")
plt.xlabel("Aire (en pixels)")
plt.ylabel("Nombre de composantes")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

cas des cellules qui se touchent et celui des petits fragments isolés.

Non, le nombre de composantes connexes ne reflète pas exactement le nombre réel de cellules :

Cellules en contact : Elles sont fusionnées en une seule composante connexe, ce qui entraîne une sous-estimation du nombre de cellules.

Fragments et bruit : Les petits artefacts isolés retenus après seuillage créent des composantes parasites, ce qui entraîne une sur-estimation.
